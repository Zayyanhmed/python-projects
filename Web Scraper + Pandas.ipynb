```python
from bs4 import BeautifulSoup

import requests
```

```python
url = "https://en.wikipedia.org/wiki/List_of_largest_companies_by_revenue"

headers = {
    "User-Agent": "ZayyanPython"
}

page = requests.get(url, headers=headers, timeout=30)

print(page.status_code)
```

```python
soup = BeautifulSoup(page.text, "html.parser")

print(soup)
```

```python
soup.find('table')
```

```python
soup.find_all ('table')[0]
```

```python
soup.find('table', class_ = 'wikitable sortable sticky-header-multi sort-under')
```

```python
table = soup.find('table', class_='wikitable sortable sticky-header-multi sort-under')
```

```python
print(table)
```

```python
import pandas as pd

world_table_titles = [
    title.get_text(" ", strip=True)
    for title in table.find("tr").find_all("th")
]

print(world_table_titles)
```

```python
df = pd.DataFrame(columns=world_table_titles)

df
```

```python
column_data = table.find_all('tr')
```

```python
for row in column_data:
    row_data = row.find_all(['th', 'td'])
    individual_row_data = [
        data.get_text(" ", strip=True) for data in row_data
    ]
    print(individual_row_data)
```

```python
import pandas as pd

tables = pd.read_html(
    "https://en.wikipedia.org/wiki/List_of_largest_companies_by_revenue",
    storage_options={"User-Agent": "ZayyanPython"},
)

print("Tables found:", len(tables))

df = tables[0].copy()

if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df = df.loc[:, ~df.columns.duplicated()]

df.columns = df.columns.str.replace(r"\[.*\]", "", regex=True).str.strip()

cols_to_drop = ["Ref.", "State-owned"]
df = df.drop(columns=[col for col in cols_to_drop if col in df.columns])

df["Name"] = df["Name"].astype(str).str.split(r"\n").str[0].str.strip()

df = df.rename(
    columns={
        "Ranks": "Rank",
        "Revenue": "Revenue (USD billions)",
        "Profit": "Profit (USD billions)",
    }
)

display(df)
print("Rows in first table:", len(df))
```
