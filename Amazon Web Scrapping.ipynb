```python
import requests
from bs4 import BeautifulSoup

URL = 'https://www.amazon.co.uk/fire-tv-stick-4k-select/dp/B0CN41YQKQ?ref=dlx_devic_dg_dcl_B0CN41YQKQ_dt_sl4_8a_pi&pf_rd_r=CG62S4M5J6BS5NZHAD4V&pf_rd_p=02f5072c-3913-4d9d-808a-2fded8461c8a&sbo=RZvfv%2F%2FHxDF%2BO5021pAnSA%3D%3D'

headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/78.0.3904.108 Safari/537.36", "Accept-Encoding":"gzip, deflate", "Accept":"text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8", "DNT":"1","Connection":"close", "Upgrade-Insecure-Requests":"1"}

page = requests.get(URL, headers=headers)

soup1 = BeautifulSoup(page.content, "html.parser")

soup2 = BeautifulSoup(soup1.prettify(), "html.parser")

title_element = soup2.find(id='productTitle')

if title_element:
    title = title_element.get_text(strip=True)
else:
    title = "Title not found"

price_element = soup2.select_one('span.a-price > span.a-offscreen')

if price_element:

    price = price_element.get_text(strip=True)

else:

    price = "Price not found"

print(f"Title: {title}")
print(f"Price: {price}")
```

```python
def check_price():
    global title, price, today

    URL = 'https://www.amazon.co.uk/fire-tv-stick-4k-select/dp/B0CN41YQKQ'

    headers = {
        "User-Agent": "Zayyan/1.0 (Macintosh; Intel Mac OS X 10_15_7) AppleWebKit/605.1.15 (KHTML, like Gecko) Version/17.0 Safari/605.1.15"
    }

    page = requests.get(URL, headers=headers, timeout=15)

    soup = BeautifulSoup(page.content, "html.parser")

    title_element = soup.find(id='productTitle')
    price_element = soup.select_one('span.a-price > span.a-offscreen')

    if not title_element or not price_element:
        print("Product title or price not found.")
        return

    title = title_element.get_text(strip=True)
    price = price_element.get_text(strip=True)
    today = datetime.date.today()

    header = ['Title', 'Price', 'Date']
    data = [title, price, today]

    file_path = Path.home() / 'Downloads' / 'AmazonWebScraperDataset.csv'

    with open(file_path, 'a', newline='', encoding='UTF-8') as f:
        writer = csv.writer(f)

        if f.tell() == 0:
            writer.writerow(header)

        writer.writerow(data)

    print(f"Title: {title}")
    print(f"Price: {price}")
    print(f"Date: {today}")
    print("Data saved to Downloads.")
```

```python
import time

while True:
    check_price()
    time.sleep(86400)
```
